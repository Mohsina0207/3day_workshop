```python
data.isnull().sum()
```

```python
#We can fill the Age column missing values with mean
data['Age'] = data['Age'].fillna(data['Age'].mean())
```

```python
data.isnull().sum()
```

```python
data['Embarked']  = data['Embarked'].fillna('S', inplace=True)
```

```python
data.drop(columns=['Cabin'],inplace=True)
```

```python
data.isnull().sum()
```

```python
#feature engineering is a process of creating,transforming or extracting imp information from the features before making it ML model ready mainly focuses on feature encoding and scaling
#Feature encoding is  a process of converting categorical values into numerical
data.head(10)
```

```python
data.columns
```

```python
#get_dummies()
pclass=pd.get_dummies(data['Pclass'],drop_first=True)
pclass
```

```python
#get_dummies()
embark=pd.get_dummies(data['Embarked'],drop_first=True,dtype=int)
embark
```

```python
sex=pd.get_dummies(data['Sex'],drop_first=True,dtype=int)
sex
```

```python
#concatenation of all tables
final_data=pd.concat([data,pclass,sex,embark],axis=1)
final_data
```

```python
final_data.drop(columns=['PassengerId','Name','Pclass','Sex','Ticket','Embarked'],inplace=True)
final_data.head()
```
